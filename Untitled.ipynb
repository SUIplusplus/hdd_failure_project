```python
import os
import requests
import zipfile
import pandas as pd
import numpy as np

# 1. Ссылки и пути для скачивания
url = "https://f001.backblazeb2.com/file/Backblaze-Hard-Drive-Data/data_Q2_2024.zip"
zip_filename = "data_Q2_2024.zip"
extract_folder = "./data/raw/data_Q2_2024"

# Скачивание файла с отслеживанием прогресса (stream=True)
if not os.path.exists(zip_filename):
    print(f"Начинаю скачивание {zip_filename} с сайта Backblaze...")
    response = requests.get(url, stream=True)
    with open(zip_filename, "wb") as f:
        for chunk in response.iter_content(chunk_size=8192):
            if chunk:
                f.write(chunk)
    print("Скачивание завершено успешно!")
else:
    print(f"Файл {zip_filename} уже скачан локально.")
```
